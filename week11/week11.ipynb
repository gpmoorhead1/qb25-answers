```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def calculate_number_of_reads(genomesize, readlength, coverage):
    return int(genomesize*coverage/readlength)

def simulate_coverage(coverage, genome_size=1000000, read_length=100):
    
    coverage_array = np.zeros(genome_size, dtype=int)
    
    num_reads = calculate_number_of_reads(genome_size, read_length, coverage)
    
    for i in range(num_reads):
        
        read_position = np.random.randint(0, genome_size-read_length+1)
        coverage_array[read_position:read_position+read_length] += 1

    return coverage_array
    
```

```python
'''
run 3x coverage
'''

genome_size = 1000000
coverage = 3
read_length = 100

genome_coverage = simulate_coverage(coverage, genome_size, read_length)
```

```python
np.savetxt('3x_coverage.txt', X=genome_coverage, delimiter=',')
```

```python
'''
code for De Bruijn Graph
'''

reads = ['ATTCA', 'ATTGA', 'CATTG', 'CTTAT', 'GATTG', 'TATTT', 'TCATT', 'TCTTA', 'TGATT', 'TTATT', 'TTCAT', 'TTCTT', 'TTGAT']

k=3

graph = set()

for read in reads:
    for i in range(len(read) - k):
        kmer1 = read[i: i+k]
        kmer2 = read[i+1: i+1+k]
        graph.add(f"{kmer1} -> {kmer2}")

with open('./debruijn_input.txt', 'w+') as f:
    f.write('digraph {\n')
    for edge in graph:
        f.write(edge)
        f.write('\n')
    f.write('}')

graph
```
